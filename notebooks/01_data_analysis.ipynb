# 01 - Business Entity Resolution: Data Analysis & EDA

This notebook performs exploratory data analysis across Source 1, Source 2, Source 3, and Ground Truth datasets.
Objectives:
1. Inspect schema, row counts, and missing value patterns.
2. Analyze entity attribute distributions (names, addresses, countries).
3. Understand match cardinality and multi-source linkage distributions in the training set.

```python
from pathlib import Path
import pandas as pd
import numpy as np

# Paths
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA_DIR = ROOT / "dataset" / "train" if (ROOT / "dataset" / "train").exists() else ROOT / "data" / "train"
print(f"Loading data from: {DATA_DIR}")
```

```python
# Load datasets
s1 = pd.read_csv(DATA_DIR / "train_source1.tsv", sep="\t", dtype=str, keep_default_na=False)
s2 = pd.read_csv(DATA_DIR / "train_source2.tsv", sep="\t", dtype=str, keep_default_na=False)
s3 = pd.read_csv(DATA_DIR / "train_source3.tsv", sep="\t", dtype=str, keep_default_na=False)
gt = pd.read_csv(DATA_DIR / "train_ground_truth.tsv", sep="\t", dtype=str, keep_default_na=False)

print(f"Source 1 shape: {s1.shape}")
print(f"Source 2 shape: {s2.shape}")
print(f"Source 3 shape: {s3.shape}")
print(f"Ground Truth shape: {gt.shape}")
```

```python
# Display sample records
display(s1.head())
display(gt.head())
```

```python
# Match statistics from ground truth
gt["match_count"] = gt["matched_entity_ids"].apply(lambda x: len(x.split(",")) if x.strip() else 0)
print("Match count distribution per S1 entity:")
print(gt["match_count"].value_counts().sort_index())
```
