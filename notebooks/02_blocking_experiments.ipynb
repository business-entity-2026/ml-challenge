# 02 - Candidate Generation & Blocking Experiments

This notebook benchmarks and evaluates candidate generation (blocking) methods.
Objectives:
1. Test rule-based and token/n-gram based blocking schemes.
2. Measure candidate reduction ratio (candidates vs. Cartesian product).
3. Evaluate candidate pair recall against train ground truth.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(ROOT / "code" / "business_entity_resolution"))

from src.load_data import load_sources, load_ground_truth
from src.preprocessing import normalize_source
from src.blocking import Blocker
```

```python
DATA_DIR = ROOT / "dataset" / "train" if (ROOT / "dataset" / "train").exists() else ROOT / "data" / "train"
s1, s2, s3 = load_sources(DATA_DIR, "train")
gt = load_ground_truth(DATA_DIR / "train_ground_truth.tsv")

# Normalize
s1, s2, s3 = map(normalize_source, (s1, s2, s3))
print(f"Loaded and normalized {len(s1)} S1, {len(s2)} S2, {len(s3)} S3 records.")
```

```python
# Run candidate generation benchmark
blocker = Blocker(name_top_k=25, address_top_k=25)
# Subset sample for fast interactive benchmarking
sample_s1 = s1.head(1000)
candidates = blocker.generate(sample_s1, s2, s3)
print(f"Generated candidates for {len(candidates)} S1 entities.")
display(candidates.head())
```
